### Handling the Missing Values


```python
# Now handling the null values of Name & Host_Name
df.dropna(subset=['name','host_name'],inplace=True)

#Now Handling the null Values of last_review & reviews_per_month
df.fillna({'last_review':0,'reviews_per_month':0},inplace=True)
```

```python
# Creating copy of Dataset For further Analysis
df_AB=df.copy()
```

```python
# Rooms where price == 0
df_AB[df_AB['price']==0].room_type.value_counts()
```

```python
# Total Rooms in dataset.
df.room_type.value_counts().sum()
```

```python
# WE can see that There are 62/225969 rooms which are price of 0.Dropping These Rooms is Reasonable Option
# It minimize the impact of your dataset and Avoid introducing imputaition errors.
df_AB=df_AB[df_AB['price']!=0]
```

### Analysis 1 >> Average Price by Neighbourhood Group.

```python
# Top 12 Neighbour_group by Avg Price.
neighbour_pivot=df_AB.pivot_table(index='neighbourhood_group',values='price',aggfunc='mean').round(2)
neighbour_pivot=neighbour_pivot.sort_values(by='price',ascending=False)
neighbour_pivot=neighbour_pivot.head(12)
neighbour_pivot
```

```python
# Plotting the Data Into Bar Chart
import matplotlib.pyplot as plt
import seaborn as sns
fig,ax=plt.subplots(1,1,figsize=(8,4))
sns.barplot(data=neighbour_pivot,x='price',y=neighbour_pivot.index,hue='price',palette='dark:r_r',legend=False,ax=ax)
sns.despine()
ax.set_xlabel('Price Of Air-Bnb',fontsize=12)
ax.set_ylabel('Neighbourhood Groups',fontsize=12)
ax.set_title(' Avg Price Of Neighbourhood Air-Bnb Groups',fontsize=14)
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=5, fontsize=10)
plt.tight_layout()
plt.show()
```

### Analysis 2 >> Most Preferred Room Type

### Analysis 3 >> Identify the most popular neighbourhoods

```python
# Identify the most popular Neighbourhoods
# Since value_counts Return the value by frequency of decending order (most to least Frequent)
top_neighbour=df.neighbourhood.value_counts().head(12)
print('The most popular Neighbourhood for Airbnb Bookings in US are:')

top_neighbour=top_neighbour.reset_index()
top_neighbour.columns=['neighbourhood','count']
top_neighbour
```

```python
# Creating Scatter plot For Most Popular Neighbourhood 
fig,ax=plt.subplots(1,1,figsize=(8,6))
sns.barplot(data=top_neighbour,x='count',y='neighbourhood',hue='count',palette='tab10',legend=False,ax=ax)
sns.despine()
ax.set_xlabel('Count of Airbnb',fontsize=12)
ax.set_ylabel('Neighbourhood Airbnb Location',fontsize=12)
ax.set_title('Count of Neighbourhood Airbnb Location in US',fontsize=14)
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=5, fontsize=10)
plt.tight_layout()
plt.show()
```

###  Analysis 4 >> Trend Of Average Listing Price Over Month

```python
# Convert last_review to_datetime format
# errors='coerce' , invalid or wrong values are converted to NAT not Time
df_AB['last_review']=pd.to_datetime(df_AB['last_review'],errors='coerce')

# Extract Month  From last_review
df_AB['month']=df_AB['last_review'].dt.month

# Calculate Average Price By month
avg_price_over_time=df_AB.groupby('month')['price'].mean()
```

```python
# Plotting the avg_price_over_time
avg_price_over_time.plot(kind='line',marker='o')
plt.title('Trend of Average Listing Price over Time',fontsize=14)
plt.xlabel('No of Month',fontsize=12)
plt.ylabel('Average Price',fontsize=12)
```

### Analysis 5 >> Price Variation by Neighbourhood Groups

```python
# Price Variation by Neighbourhood Groups
price_variation=df_AB.groupby('neighbourhood_group')['price'].agg('sum')
price_variation=price_variation.reset_index()
price_variation.columns=['neighbourhood_groups','sum']
price_variation=price_variation.sort_values(by='sum',ascending=False).head(8)
price_variation
```

```python
# Plotting The Price variation By NEighbourhood Groups.
from matplotlib.ticker import FuncFormatter
fig,ax=plt.subplots(1,1,figsize=(6,5))
sns.scatterplot(data=price_variation,x='neighbourhood_groups',y='sum')
ax.set_title('Price Variation By Neghbourhood Groups ',fontsize=16)
ax.set_xlabel('Neighbourhood Groups',fontsize=12)
ax.set_ylabel('Sum Of Price',fontsize=12)
ax.tick_params(axis='x', rotation=45)
ax.yaxis.set_major_formatter(
    FuncFormatter(lambda x,pos:f'{x/1000:.0f}K')
)
sns.despine()
plt.tight_layout()
plt.show()
```

###  Analysis 6  >> Price VS Number of Reviews

```python
sns.scatterplot(data=df_AB,x='number_of_reviews',y='price',hue='price',palette='viridis',legend=False)
plt.title('Price by Number of Reviews',fontsize=15)
plt.xlabel('Number of Reviews',fontsize=12)
plt.ylabel('Price Of Airbnb',fontsize=12)
sns.despine()
plt.tight_layout()
plt.show()
```

### Analysis 7 >> Number of Hosts By Neighbourhood Groups And Room Type

```python
# Number of Hosts by Neighbourhood Groups and Room Type.

# Calculate the number of Hosts for each combination of neighbourhood group and room type.
hosts=df_AB.pivot_table(index=['neighbourhood_group','room_type'],values='host_id',aggfunc='nunique')
hosts=hosts.reset_index()
hosts=hosts.head(20)
```

```python
# Plotting the barchart for Number of hosts for each neighourhood groups and room type
fig,ax=plt.subplots(1,1,figsize=(8,4))
sns.barplot(data=hosts,x='neighbourhood_group',y='host_id',hue='room_type',ax=ax)
ax.legend(title='Room Type')
ax.set_xlabel('Neighbourhood Groups & Room Type',fontsize=12)
ax.set_ylabel(' Number Of Hosts ',fontsize=12)
ax.set_title('Number of Hosts by Neighbourhood Groups & Room Type',fontsize=14)
plt.tight_layout()
plt.show()
```

### Additional - Chart Correlation Heatmap

```python
# Correlation Heatmap Visualization Code.

# Drop the Specific column That Is not Useful For this Heatmap.
df_numerical = df_AB.select_dtypes(include='number')
# Compute the correlation Matrix.
corr_matrix=df_numerical.corr()

# Create A Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix,annot=True,cmap='magma')
plt.title('Correlation Heatmap of Selected Numerical Values',fontsize=16)
plt.show()
```
